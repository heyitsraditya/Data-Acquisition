## Tugas 3 — Paragraf reproducibility

**Pentingnya Reproducibility dalam Akuisisi Data dan contoh kasus nyatanya**

Reproducibility atau keterulangan merupakan prinsip fundamental dalam akuisisi dan manajemen data yang menjamin bahwa seluruh alur pengumpulan data dapat dijalankan kembali oleh peneliti lain dengan menghasilkan luaran (output) yang identik secara konsisten. Dalam siklus hidup data science, menjaga keterulangan proses akuisisi sangat krusial untuk menjamin integritas, transparansi, keandalan, serta validitas data sebelum dilanjutkan ke tahap analisis dan pemodelan lanjutan. Tanpa skrip otomatisasi yang terstruktur, dokumentasi parameter yang jelas, serta penerapan kontrol versi (version control), data mentah yang diunduh dari internet atau API dapat berubah tanpa terdeteksi, sehingga menyebabkan analisis berikutnya bersifat bias, tidak konsisten, dan sangat sulit untuk diverifikasi kebenarannya.

Sebagai contoh kasus nyata, pada riset pemantauan tren harga produk secara berkala melalui teknik web scraping dari platform e-commerce, struktur kode HTML halaman web dapat berubah sewaktu-waktu tanpa pemberitahuan. Jika alur akuisisi data tidak menerapkan prinsip reproducibility—seperti menyimpan berkas mentah asli bersifat read-only di folder data/raw dan mencatat versi skrip scraper—tim data tidak dapat memverifikasi apakah perubahan lonjakan harga disebabkan oleh dinamika pasar asli atau kesalahan pengambilan data. Oleh karena itu, standarisasi struktur proyek dan dokumentasi yang konsisten menjadi kunci utama keandalan data science.
